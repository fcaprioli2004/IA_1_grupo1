# Ejercicios Teóricos

### 1. Defina con sus propias palabras inteligencia natural, inteligencia artificial y agente.

**Inteligencia natural**: es la capacidad que poseen los seres vivos, especialmente las personas, para percibir su entorno, aprender de la experiencia, razonar, resolver problemas, adaptarse y tomar decisiones.

**Inteligencia artificial**: es la capacidad de una máquina o sistema informático para realizar tareas que normalmente requieren inteligencia, como aprender, razonar, reconocer patrones, tomar decisiones o resolver problemas.

**Agente**: es una entidad capaz de percibir lo que ocurre en su entorno, procesar esa información y realizar acciones para alcanzar un determinado objetivo.


### 2. ¿Qué es un agente racional?

Un agente es aquel capaz de tomar informacion del medio ambiente(sensar) y tomar una desicion para modificar el mismo (actuar). Es racional, si la acción maximiza el exito frente a las condiciones que se encuentra

### 3. ¿Un agente es siempre una computadora?

No, un agente no es necesariamente una computadora. Puede ser una persona, un mecanismo, un robot o un sistema de software, siempre que pueda percibir su entorno y realizar acciones sobre él para alcanzar un objetivo.

### 4. Defina Omnisciencia, Aprendizaje y Autonomía.

1. **Omnisciencia:** es la capacidad de conocer todo, o al menos toda la información necesaria para tomar una decisión correcta.

2. **Aprendizaje:** es la capacidad de utilizar las experiencias y percepciones obtenidas para mejorar el comportamiento y tomar mejores decisiones en el futuro.

3. **Autonomía:** es la capacidad de actuar basándose en la propia experiencia, aprendiendo y compensando la falta de conocimiento inicial sin depender completamente de información previamente incorporada.

## 5. Tipos de agentes según su estructura

1. **Agente reactivo simple:**  
   Actúa únicamente según lo que percibe en ese momento. Utiliza reglas del tipo **“si ocurre esto, hacer aquello”**, sin tener en cuenta situaciones anteriores.  
   **Ejemplo:** una aspiradora que, si detecta suciedad, aspira; y si el lugar está limpio, se desplaza.

2. **Agente reactivo basado en modelos:**  
   Además de la percepción actual, mantiene un **estado interno o modelo del entorno**. Esto le permite tener en cuenta aspectos que no puede observar directamente en ese instante. Suele estar desarrollado de antemano la forma de relizar cambios mediante algun modelo matemático o algo similar, a diferencia de el que aprende que modifica su comportamiento a partir de sus propias experiencias previas.
   
   **Ejemplo:** un termostato inteligente que considera la temperatura actual y un modelo de cómo evoluciona la temperatura de la habitación para decidir cuándo encender o apagar la calefacción.

3. **Agente basado en objetivos:**  
   Utiliza un modelo del entorno y además posee un **objetivo que desea alcanzar**. Antes de actuar, puede analizar distintas acciones y elegir aquellas que lo acerquen a su meta.  
   **Ejemplo:** un sistema de navegación GPS que calcula el camino mas corto a seguir para llegar a un destino.

4. **Agente basado en utilidad:**  
   Además de tener objetivos, utiliza una **función de utilidad** para comparar diferentes resultados y elegir el que resulte más conveniente. No solo busca cumplir el objetivo, sino hacerlo de la mejor manera posible.  
   **Ejemplo:** un sistema de climatización que busca mantener una temperatura confortable pero, al mismo tiempo, reducir el consumo de energía.

5. **Agente que aprende:**  
   Tiene la capacidad de **adaptar su comportamiento a partir de sus experiencias previas**. Utiliza la información obtenida de sus acciones y resultados para modificar su forma de actuar en el futuro.  
   **Ejemplo:** un robot que aprende progresivamente cuál es la mejor ruta para desplazarse evitando obstáculos.

### 6. Para los siguientes entornos de trabajo indique sus **propiedades**:

**a. Una partida de ajedrez**

  -Observabilidad: Totalmente observable (ambos jugadores ven todas las
piezas y el estado completo del tablero).
  
  -Determinismo: Determinista (no hay azar en los movimientos; las piezas se
mueven de forma exacta y predecible).
  
  -Secuencia: Secuencial (cada movimiento afecta estratégicamente a todos los
movimientos futuros).
  
  -Dinámica: Estático (el entorno no cambia mientras el jugador está pensando
su siguiente movimiento).
  
  -Naturaleza: Discreto (el tablero tiene un número finito de casillas y
estados posibles).
  
  -Agentes: Multiagente (juegas contra otro agente).

**b. Un partido de baloncesto**

  -Observabilidad: Parcialmente observable (un jugador no puede ver todo lo que ocurre a su espalda ni conocer las jugadas planificadas del equipo contrario).

  -Determinismo: Estocástico (el resultado de un tiro o un pase depende de factores físicos impredecibles, el viento, el rebote, etc).

  -Secuencia: Secuencial (las acciones de los primeros cuartos afectan el cansancio y el marcador para el final del partido).

  -Dinámica: Dinámico (el entorno cambia constantemente independientemente de si un jugador se detiene a pensar).

  -Naturaleza: Continuo (los movimientos se dan en un espacio físico continuo y en tiempo real).

  -Agentes: Multiagente (cooperativo con tu equipo y competitivo contra el rival).

**c. El juego Pacman**

  -Observabilidad: Totalmente observable (el jugador ve todo el laberinto).

  -Determinismo: Determinista (el comportamiento de los fantasmas responde a un algoritmo fijo y predecible, a veces se perciba como aleatorio).

  -Secuencia: Secuencial (comer pastillas ahora afecta las rutas y opciones de supervivencia más adelante).

  -Dinámica: Dinámico (los fantasmas se mueven todo el tiempo).

  -Naturaleza: Discreto (las decisiones son puntuales: arriba, abajo, izquierda, derecha).

  -Agentes: Individual (los fantasmas no pueden modificar el ambiente, responden a un algoritmo).

**d. El truco**

  -Observabilidad: Parcialmente observable (no puedes ver las cartas de tus oponentes ni las que quedan en el mazo).

  -Determinismo: Estocástico (el reparto inicial de las cartas introduce un elemento de azar fundamental).

  -Secuencia: Secuencial / Episódico (es secuencial dentro de la misma mano porque las cartas jugadas afectan esa ronda, pero es episódico entre distintas manos ya que cada nueva ronda comienza desde cero).

  -Dinámica: Estático (se pude tomar el tiempo necesario para tomar una decisión).

  -Naturaleza: Discreto (cartas, puntos y turnos finitos y contables).

  -Agentes: Multiagente (competitivo y cooperativo si se juega en parejas).

**e. Las damas**

  -Observabilidad: Totalmente observable (todo el tablero está a la vista).

  -Determinismo: Determinista (las reglas y movimientos no tienen variables de azar).

  -Secuencia: Secuencial (cada captura o movimiento define el resto de la partida).

  -Dinámica: Estático (el entorno no cambia mientras es tu turno).

  -Naturaleza: Discreto (casillas y fichas limitadas).

  -Agentes: Multiagente (competitivo).

**f. El juego tres en raya (Ta-te-ti)**

  -Observabilidad: Totalmente observable (cuadrícula de 3x3 visible para ambos).

  -Determinismo: Determinista (sin dados ni elementos aleatorios).

  -Secuencia: Secuencial (poner una marca bloquea casillas para el futuro).

  -Dinámica: Estático (el tablero espera tu movimiento).

  -Naturaleza: Discreto (solo 9 posiciones posibles).

  -Agentes: Multiagente (competitivo).

**g. Un jugador de Pokémon Go**

  -Observabilidad: Parcialmente observable (el jugador no sabe qué Pokémon aparecerán cerca, ni qué hacen otros jugadores en gimnasios lejanos).

  -Determinismo: Estocástico (las tasas de aparición, la probabilidad de captura y el comportamiento del Pokémon tienen un fuerte componente de azar).

  -Secuencia: Secuencial (gastar recursos como Pokéballs ahora afecta tu capacidad de capturar más tarde).

  -Dinámica: Dinámico (el mundo real cambia, los Pokémon desaparecen por tiempo y los gimnasios son atacados en tiempo real).

  -Naturaleza: Continuo (tu ubicación se basa en coordenadas GPS continuas).

  -Agentes: Multiagente (interactúas y compites por recursos o gimnasios con otros jugadores a nivel global).

**h. Un robot explorador autónomo de Marte**

  -Observabilidad: Parcialmente observable (sus sensores tienen alcance limitado, hay rocas que bloquean la visión y las cámaras pueden ensuciarse).

  -Determinismo: Estocástico (el terreno puede ceder, las ruedas pueden resbalar y el clima marciano es impredecible).

  -Secuencia: Secuencial (decidir avanzar por un terreno difícil afecta los niveles de batería y la integridad mecánica para el resto de la misión).

  -Dinámica: Dinámico (el entorno cambia independientemente del robot; el sol se mueve afectando los paneles solares, hay tormentas de arena, etc.).

  -Naturaleza: Continuo (se mueve por un espacio físico con ángulos de giro, velocidades y voltajes de batería que no son números enteros).

  -Agentes: Agente individual (generalmente opera solo en su entorno inmediato, a menos que esté programado para coordinarse en enjambre con otros rovers).

#### d. El truco

Consideramos como agente a **uno de los jugadores de truco**. Suponemos que cada jugador únicamente puede observar **sus propias cartas**, además de las cartas que ya fueron jugadas y los cantos realizados. No puede observar las cartas de los demás jugadores, por lo que debe tomar decisiones utilizando información incompleta.

- **Parcialmente observable:** no se conocen las cartas de los demás jugadores.
- **Estocástico:** el reparto de cartas introduce azar y existe información desconocida.
- **Secuencial:** cada carta jugada o canto realizado afecta las decisiones posteriores.
- **Estático:** mientras un jugador decide, suponemos que el estado del juego no cambia.
- **Discreto:** existe una cantidad definida de cartas y acciones posibles.
- **Multiagente:** participan varios jugadores con objetivos enfrentados.

Para cada caso consideramos **R = Rendimiento, E = Entorno, A = Actuadores y S = Sensores**.

---

#### a. Crucigrama

Consideramos como agente a un **sistema encargado de resolver un crucigrama**. El agente observa las pistas, la cantidad de casillas y las letras que ya fueron colocadas. A partir de esa información decide qué palabras escribir o modificar.

| REAS | Descripción |
|---|---|
| **R - Rendimiento** | Completar correctamente la mayor cantidad posible de palabras, minimizar errores y, si se considera, resolverlo en el menor tiempo posible. |
| **E - Entorno** | Grilla del crucigrama, casillas, pistas horizontales y verticales y letras ya colocadas. |
| **A - Actuadores** | Escribir letras o palabras, borrar y reemplazar respuestas. |
| **S - Sensores** | Leer las pistas, observar las casillas disponibles y reconocer las letras ya ingresadas. |

---

#### b. Taxi circulando

Consideramos como agente a un **taxi autónomo** que debe trasladar pasajeros desde un punto de origen hasta un destino. Debe observar continuamente el tránsito y las condiciones del camino para decidir cómo conducir de manera segura y eficiente.

| REAS | Descripción |
|---|---|
| **R - Rendimiento** | Llegar al destino de manera segura, respetar las normas de tránsito, reducir el tiempo de viaje y el consumo de combustible, y mantener el confort del pasajero. |
| **E - Entorno** | Calles, rutas, otros vehículos, peatones, pasajeros, semáforos, señales de tránsito, obstáculos y condiciones climáticas. |
| **A - Actuadores** | Volante, acelerador, freno, caja de cambios, luces, bocina y señalizadores. |
| **S - Sensores** | Cámaras, GPS, sensores de distancia, radar o LiDAR, velocímetro y sensores del vehículo. |

---

#### c. Robot clasificador de piezas

Consideramos como agente a un **robot encargado de identificar y separar piezas** que llegan, por ejemplo, a través de una cinta transportadora. El robot debe reconocer las características de cada pieza y colocarla en el lugar correspondiente.

| REAS | Descripción |
|---|---|
| **R - Rendimiento** | Clasificar correctamente las piezas, reducir errores, evitar daños y realizar el proceso en el menor tiempo posible. |
| **E - Entorno** | Cinta transportadora, piezas a clasificar, zona de trabajo y recipientes o sectores donde se depositan las piezas. |
| **A - Actuadores** | Brazo robótico, motores, pinza o sistema de agarre y mecanismos para desplazar las piezas. |
| **S - Sensores** | Cámaras, sensores de presencia, color, forma, tamaño, posición u otros sensores necesarios para identificar las piezas. |
