# Ejercicios Teóricos

### 1. Defina con sus propias palabras inteligencia natural, inteligencia artificial y agente.

**Inteligencia natural**: es la capacidad que poseen los seres vivos, especialmente las personas, para percibir su entorno, aprender de la experiencia, razonar, resolver problemas, adaptarse y tomar decisiones.

**Inteligencia artificial**: es la capacidad de una máquina o sistema informático para realizar tareas que normalmente requieren inteligencia, como aprender, razonar, reconocer patrones, tomar decisiones o resolver problemas.

**Agente**: es una entidad capaz de percibir lo que ocurre en su entorno, procesar esa información y realizar acciones para alcanzar un determinado objetivo.


### 2. ¿Qué es un agente racional?

Un agente es aquel capaz de tomar informacion del medio ambiente(sensar) y tomar una desicion para modificar el mismo (actuar). Es racional, si la acción maximiza el exito frente a las condiciones que se encuentra

### 3. ¿Un agente es siempre una computadora?

No, un agente no es necesariamente una computadora. Puede ser una persona, un mecanismo, un robot o un sistema de software, siempre que pueda percibir su entorno y realizar acciones sobre él para alcanzar un objetivo.

### 4. Defina Omnisciencia, Aprendizaje y Autonomía.

1. **Omnisciencia:** es la capacidad de conocer todo, o al menos toda la información necesaria para tomar una decisión correcta.

2. **Aprendizaje:** es la capacidad de utilizar las experiencias y percepciones obtenidas para mejorar el comportamiento y tomar mejores decisiones en el futuro.

3. **Autonomía:** es la capacidad de actuar basándose en la propia experiencia, aprendiendo y compensando la falta de conocimiento inicial sin depender completamente de información previamente incorporada.

## 5. Tipos de agentes según su estructura

1. **Agente reactivo simple:**  
   Actúa únicamente según lo que percibe en ese momento. Utiliza reglas del tipo **“si ocurre esto, hacer aquello”**, sin tener en cuenta situaciones anteriores.  
   **Ejemplo:** una aspiradora que, si detecta suciedad, aspira; y si el lugar está limpio, se desplaza.

2. **Agente reactivo basado en modelos:**  
   Además de la percepción actual, mantiene un **estado interno o modelo del entorno**. Esto le permite tener en cuenta aspectos que no puede observar directamente en ese instante. Suele estar desarrollado de antemano la forma de relizar cambios mediante algun modelo matemático o algo similar, a diferencia de el que aprende que modifica su comportamiento a partir de sus propias experiencias previas.
   
   **Ejemplo:** un termostato inteligente que considera la temperatura actual y un modelo de cómo evoluciona la temperatura de la habitación para decidir cuándo encender o apagar la calefacción.

3. **Agente basado en objetivos:**  
   Utiliza un modelo del entorno y además posee un **objetivo que desea alcanzar**. Antes de actuar, puede analizar distintas acciones y elegir aquellas que lo acerquen a su meta.  
   **Ejemplo:** un sistema de navegación GPS que calcula el camino mas corto a seguir para llegar a un destino.

4. **Agente basado en utilidad:**  
   Además de tener objetivos, utiliza una **función de utilidad** para comparar diferentes resultados y elegir el que resulte más conveniente. No solo busca cumplir el objetivo, sino hacerlo de la mejor manera posible.  
   **Ejemplo:** un sistema de climatización que busca mantener una temperatura confortable pero, al mismo tiempo, reducir el consumo de energía.

5. **Agente que aprende:**  
   Tiene la capacidad de **adaptar su comportamiento a partir de sus experiencias previas**. Utiliza la información obtenida de sus acciones y resultados para modificar su forma de actuar en el futuro.  
   **Ejemplo:** un robot que aprende progresivamente cuál es la mejor ruta para desplazarse evitando obstáculos.

8. La Hormiga de Langton es un agente capaz de modificar el estado de la casilla en la que se encuentra para colorearla o bien de blanco o de negro. Al comenzar, la ubicación de la hormiga es una casilla aleatoria y mira hacia una de las cuatro casillas adyacentes. Si...

# Caracterización de la Hormiga de Langton y Análisis de Patrones

---

## 1. Tabla REAS (PEAS) del Agente

Para caracterizar a la Hormiga de Langton como un **agente reflexivo simple**, definimos sus cuatro elementos fundamentales:

| Componente REAS | Descripción |
| --- | --- |
| **Rendimiento** *(Performance)* | Modificación y cobertura correcta del espacio de la cuadrícula siguiendo estrictamente las reglas de transición (cambio de color y giro). |
| **Entorno** *(Environment)* | Cuadrícula bidimensional (infinita o delimitada) compuesta por casillas discretas en uno de dos estados posibles: **blanco** o **negro**. |
| **Actuadores** *(Actuators)* | **1.** Mecanismo para cambiar el color de la casilla actual (blanco $\leftrightarrow$ negro).<br><br>**2.** Mecanismo de rotación (girar 90° a la derecha o a la izquierda).<br><br>**3.** Mecanismo de desplazamiento (avanzar una casilla en la dirección actual). |
| **Sensores** *(Sensors)* | Sensor de color/estado para detectar la casilla actual en la que se encuentra posicionado el agente, junto con el registro de su orientación actual (Norte, Sur, Este, Oeste). |

---

## 2. Propiedades del Entorno

* **Parcialmente observable (localmente):** La hormiga solo percibe el estado de la casilla en la que está ubicada en ese preciso instante; no posee una vista global del tablero.
* **Determinista:** El estado siguiente de la casilla, la orientación y la posición de la hormiga están 100% determinados por las reglas fijas y el estado actual; no interviene el azar.
* **Secuencial:** Cada movimiento y cambio de color altera la configuración del entorno, condicionando los pasos futuros cuando la hormiga vuelva a transitar por esa zona.
* **Estático:** El entorno no cambia de manera autónoma; únicamente la hormiga altera el estado de la cuadrícula.
* **Discreto:** La posición (coordenadas en la matriz), los estados (blanco/negro), las direcciones posibles (4) y el tiempo (iteraciones) son finitos y discontinuos.
* **Agente individual:** La hormiga opera de manera aislada sin la interacción, competencia ni cooperación de otros agentes.

---

## 3. Patrones Emergentes

Al ejecutar el algoritmo comenzando en un tablero completamente blanco, el comportamiento de la hormiga atraviesa **tres fases bien definidas**:

1. **Fase de Simetría Inicial (pasos 0 a ~500):** La hormiga genera patrones geométricos relativamente simples y notablemente simétricos.
2. **Fase Caótica (pasos ~500 a 10,287):** Se rompe la simetría y el agente entra en un comportamiento pseudoaleatorio, construyendo una "mancha" desordenada y densa de casillas negras y blancas.
3. **Fase Periódica / "La Carretera" (a partir del paso 10,288):**

> **Patrón Repetitivo:**
> A partir de la **iteración 10,288**, la hormiga entra en un bucle periódico exacto de **104 pasos**. Desde este momento, la hormiga repite de manera infinita esa misma secuencia de 104 movimientos, construyendo una estructura diagonal constante conocida como **"la carretera"** (*highway*).

*(Nota: La **Conjetura de la Hormiga de Langton** sostiene que, independientemente de la configuración inicial finita del tablero, la hormiga siempre terminará eventualmente generando esta "carretera").*

```python
import pygame
import sys

#Configuración de la ventana
WIDTH, HEIGHT = 1000, 800  # Tamaño de la ventana (pixeles)
CELL_SIZE = 5            # tamaño de cada celda (pixeles)
# Velocidad de simulación
STEPS_PER_FRAME = 8     

# Colores (R, G, B)
BG_COLOR = (240, 240, 240)  # Color del fondo
BLACK_CELL = (30, 30, 30)  # Color del camino
ANT_COLOR = (255, 69, 0)   # Color de la hormiga

# Vectores de dirección para la hormiga: Norte, Este, Sur, Oeste
# En pygame, el eje Y crece hacia abajo, por lo que Norte es (0, -1)
DIRECTIONS = [
    (0, -1),  # North
    (1, 0),   # East
    (0, 1),   # South
    (-1, 0)   # West
]

# --- Inicio de Pygame ---
pygame.init()
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Hormiga de Langton")
clock = pygame.time.Clock()

# --- Simulacion ---
# Al principio, es todo blanco, así que solo necesitamos almacenar las celdas negras.
black_cells = set()

# Se definen las condiciones iniciales de la hormiga (posición y dirección)
ant_x, ant_y = 0, 0
ant_direction = 3  

# Bucle de juego
running = True
steps = 0
paused = False

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN:
            if event.key == pygame.K_SPACE:
                paused = not paused
                if paused:
                    pygame.display.set_caption("Langton's Ant - PAUSED")
                else:
                    pygame.display.set_caption(f"Langton's Ant - Steps: {steps}")

    if paused:
        continue

    # Este for admite la posibilidad de ejecutar múltiples pasos de la simulación por cada frame renderizado, lo que permite una simulación más rápida.
    for _ in range(STEPS_PER_FRAME):
        current_pos = (ant_x, ant_y)
        
        # Lógica de la hormiga: según el color de la celda actual, gira y cambia el color de la celda.
        if current_pos in black_cells:
            # Si la celda es negra: Gira 90 a la izquierda, cambia el color de la celda a blanca
            ant_direction = (ant_direction - 1) % 4
            black_cells.remove(current_pos)
        else:
            # Si la celda es blanca: Gira 90 a la derecha, cambia el color de la celda a negra
            ant_direction = (ant_direction + 1) % 4
            black_cells.add(current_pos)

        # Mover la hormiga a la siguiente celda según su dirección actual
        dx, dy = DIRECTIONS[ant_direction]
        ant_x += dx
        ant_y += dy
        steps += 1

    # 4.Renderizado
    screen.fill(BG_COLOR)  # Resetea la pantalla a blanco antes de dibujar las celdas negras y la hormiga.

    # Se centra el (0,0)
    center_x = WIDTH // 2
    center_y = HEIGHT // 2

    # Se dibujan todas las celdas negras almacenadas en black_cells
    for bx, by in black_cells:
        # Convierte las coordenadas de la celda a coordenadas de pantalla
        rect_x = center_x + (bx * CELL_SIZE)
        rect_y = center_y + (by * CELL_SIZE)
        
        # Dibuja el rectángulo negro correspondiente a la celda
        cell_rect = pygame.Rect(rect_x, rect_y, CELL_SIZE, CELL_SIZE)
        pygame.draw.rect(screen, BLACK_CELL, cell_rect)

    # Dibuja la posición actual de la hormiga (marcador opcional, útil para ver su ubicación)
    ant_rect_x = center_x + (ant_x * CELL_SIZE)
    ant_rect_y = center_y + (ant_y * CELL_SIZE)
    ant_rect = pygame.Rect(ant_rect_x, ant_rect_y, CELL_SIZE, CELL_SIZE)
    pygame.draw.rect(screen, ANT_COLOR, ant_rect)

    # 5. Display Flip and Framerate control
    pygame.display.flip()
    
    # Se establecen los fotogramas por segundo
    clock.tick(60) 
    
    # Update title bar info every frame (without flickering title too fast)
    if steps % (STEPS_PER_FRAME * 2) == 0:
        pygame.display.set_caption(f"Hormiga de Langton - Pasos: {steps} | Celdas: {len(black_cells)}")

# Se cierra el juego
pygame.quit()
sys.exit()
```

9. El Juego de la Vida de Conway consiste en un tablero donde cada casilla representa una célula, de manera que a cada célula le rodean 8 vecinas. Las células tienen dos estados: están *vivas* o *muertas*. En cada iteración, el estado de todas las células se tiene en cuenta para calcular el estado siguiente en simultáneo de acuerdo a las siguientes acciones:

* Nacer: Si una célula muerta tiene exactamente 3 células vecinas vivas, dicha célula pasa a estar viva.

* Morir: Una célula viva puede morir sobrepoblación cuando tiene más de tres vecinos alrededor o por aislamiento si tiene solo un vecino o ninguno.

* Vivir: una célula se mantiene viva si tiene 2 o 3 vecinos a su alrededor.

    Caracterice el agente con su tabla REAS y las propiedades del entorno para después programarlo en Python:

## Caracterización del agente

Consideramos como agente al **sistema encargado de actualizar el tablero completo**.

En cada generación, el agente observa el estado actual de las células y, para cada una de ellas, determina la cantidad de vecinos vivos. A partir de esta información aplica las reglas del Juego de la Vida para obtener el estado correspondiente en la siguiente generación.

Es importante que todas las células sean actualizadas de manera simultánea. Por este motivo, primero calculamos completamente la nueva generación y luego reemplazamos el tablero anterior.

## Tabla REAS

| Elemento | Descripción |
|---|---|
| **R - Rendimiento** | Aplicar correctamente las reglas del Juego de la Vida y representar correctamente la evolución del tablero. |
| **E - Entorno** | Tablero bidimensional formado por células vivas y muertas. |
| **A - Actuadores** | Cambiar el estado de las células entre viva y muerta para generar la siguiente generación. |
| **S - Sensores** | Conocer el estado actual de cada célula y el estado de sus células vecinas. |

## Propiedades del entorno

Consideramos que el agente tiene acceso al estado completo del tablero antes de calcular cada nueva generación.

- **Totalmente observable:** el agente conoce el estado de todas las células del tablero.
- **Determinista:** para un mismo estado inicial, siempre se obtiene la misma evolución.
- **Secuencial:** cada generación depende del estado obtenido en la generación anterior.
- **Estático:** el tablero no cambia mientras se está calculando la siguiente generación.
- **Discreto:** las posiciones del tablero y los estados de las células son discretos.
- **Agente individual:** consideramos al sistema completo encargado de actualizar el tablero como un único agente.

```python
import pygame
import random


# ============================================================
# CONFIGURACIÓN
# ============================================================

# Tamaño de la ventana en píxeles.
WIDTH = 1000
HEIGHT = 800

# Tamaño de cada célula en píxeles.
CELL_SIZE = 10

# Cantidad de filas y columnas del tablero.
COLUMNAS = WIDTH // CELL_SIZE
FILAS = HEIGHT // CELL_SIZE

# Velocidad de la simulación.
FPS = 10

# Probabilidad inicial de que una célula esté viva.
PROBABILIDAD_VIVA = 0.25


# ============================================================
# COLORES
# ============================================================

COLOR_FONDO = (240, 240, 240)
COLOR_CELULA_VIVA = (30, 30, 30)
COLOR_GRILLA = (210, 210, 210)


# ============================================================
# CREACIÓN DEL TABLERO
# ============================================================

def crear_tablero():
    """
    Crea un tablero con células vivas y muertas de forma aleatoria.

    0 = célula muerta
    1 = célula viva
    """

    tablero = []

    for fila in range(FILAS):

        nueva_fila = []

        for columna in range(COLUMNAS):

            if random.random() < PROBABILIDAD_VIVA:
                nueva_fila.append(1)
            else:
                nueva_fila.append(0)

        tablero.append(nueva_fila)

    return tablero


# ============================================================
# CONTEO DE VECINOS
# ============================================================

def contar_vecinos(tablero, fila, columna):
    """
    Cuenta cuántas células vivas rodean a una célula.

    Se analizan las ocho posiciones vecinas.
    Las posiciones fuera del tablero se consideran muertas.
    """

    vecinos_vivos = 0

    # Recorremos las ocho posiciones que rodean a la célula.
    for desplazamiento_fila in (-1, 0, 1):
        for desplazamiento_columna in (-1, 0, 1):

            # No contamos la propia célula.
            if desplazamiento_fila == 0 and desplazamiento_columna == 0:
                continue

            fila_vecina = fila + desplazamiento_fila
            columna_vecina = columna + desplazamiento_columna

            # Verificamos que la posición esté dentro del tablero.
            if (
                0 <= fila_vecina < FILAS
                and 0 <= columna_vecina < COLUMNAS
            ):
                vecinos_vivos += tablero[fila_vecina][columna_vecina]

    return vecinos_vivos


# ============================================================
# SIGUIENTE GENERACIÓN
# ============================================================

def siguiente_generacion(tablero):
    """
    Calcula el estado completo de la siguiente generación.

    Se utiliza un tablero nuevo para que todas las células
    cambien de estado de manera simultánea.
    """

    # Inicialmente todas las células de la nueva generación están muertas.
    nuevo_tablero = [
        [0 for columna in range(COLUMNAS)]
        for fila in range(FILAS)
    ]

    for fila in range(FILAS):
        for columna in range(COLUMNAS):

            vecinos_vivos = contar_vecinos(tablero, fila, columna)
            celula_viva = tablero[fila][columna] == 1

            # Una célula viva sobrevive si tiene 2 o 3 vecinos vivos.
            if celula_viva and vecinos_vivos in (2, 3):
                nuevo_tablero[fila][columna] = 1

            # Una célula muerta nace si tiene exactamente 3 vecinos vivos.
            elif not celula_viva and vecinos_vivos == 3:
                nuevo_tablero[fila][columna] = 1

            # En cualquier otro caso la célula queda muerta.

    return nuevo_tablero


# ============================================================
# DIBUJO DEL TABLERO
# ============================================================

def dibujar_tablero(pantalla, tablero):
    """
    Dibuja las células vivas y la grilla sobre la pantalla.
    """

    pantalla.fill(COLOR_FONDO)

    # Dibujar las células vivas.
    for fila in range(FILAS):
        for columna in range(COLUMNAS):

            if tablero[fila][columna] == 1:

                x = columna * CELL_SIZE
                y = fila * CELL_SIZE

                celda = pygame.Rect(
                    x,
                    y,
                    CELL_SIZE,
                    CELL_SIZE
                )

                pygame.draw.rect(
                    pantalla,
                    COLOR_CELULA_VIVA,
                    celda
                )

    # Dibujar líneas verticales.
    for x in range(0, WIDTH, CELL_SIZE):
        pygame.draw.line(
            pantalla,
            COLOR_GRILLA,
            (x, 0),
            (x, HEIGHT)
        )

    # Dibujar líneas horizontales.
    for y in range(0, HEIGHT, CELL_SIZE):
        pygame.draw.line(
            pantalla,
            COLOR_GRILLA,
            (0, y),
            (WIDTH, y)
        )


# ============================================================
# INICIO DE PYGAME
# ============================================================

pygame.init()

pantalla = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Juego de la Vida de Conway")

reloj = pygame.time.Clock()


# ============================================================
# ESTADO INICIAL
# ============================================================

tablero = crear_tablero()

ejecutando = True
pausado = False
generacion = 0


# ============================================================
# BUCLE PRINCIPAL
# ============================================================

while ejecutando:

    # --------------------------------------------------------
    # Eventos
    # --------------------------------------------------------

    for evento in pygame.event.get():

        # Cerrar la ventana.
        if evento.type == pygame.QUIT:
            ejecutando = False

        # Controles mediante teclado.
        elif evento.type == pygame.KEYDOWN:

            # ESPACIO: pausar o continuar.
            if evento.key == pygame.K_SPACE:
                pausado = not pausado

            # R: crear un tablero aleatorio nuevo.
            elif evento.key == pygame.K_r:
                tablero = crear_tablero()
                generacion = 0

            # C: limpiar completamente el tablero.
            elif evento.key == pygame.K_c:

                tablero = [
                    [0 for columna in range(COLUMNAS)]
                    for fila in range(FILAS)
                ]

                generacion = 0

            # N: avanzar una generación manualmente si está pausado.
            elif evento.key == pygame.K_n and pausado:

                tablero = siguiente_generacion(tablero)
                generacion += 1

        # ----------------------------------------------------
        # Modificar células con el mouse
        # ----------------------------------------------------

        elif evento.type == pygame.MOUSEBUTTONDOWN:

            mouse_x, mouse_y = pygame.mouse.get_pos()

            columna = mouse_x // CELL_SIZE
            fila = mouse_y // CELL_SIZE

            # Cambiar el estado de la célula seleccionada.
            tablero[fila][columna] = 1 - tablero[fila][columna]


    # --------------------------------------------------------
    # Actualizar simulación
    # --------------------------------------------------------

    if not pausado:

        tablero = siguiente_generacion(tablero)
        generacion += 1


    # --------------------------------------------------------
    # Renderizado
    # --------------------------------------------------------

    dibujar_tablero(pantalla, tablero)


    # Actualizar el título de la ventana.
    if pausado:
        estado = "PAUSADO"
    else:
        estado = "EJECUTANDO"

    pygame.display.set_caption(
        f"Juego de la Vida - Generación: {generacion} | {estado}"
    )


    # Mostrar el nuevo frame.
    pygame.display.flip()


    # Limitar la velocidad de la simulación.
    reloj.tick(FPS)


# ============================================================
# CIERRE
# ============================================================

pygame.quit()
```
